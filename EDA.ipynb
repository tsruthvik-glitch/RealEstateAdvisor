```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("✅ Core libraries loaded successfully")
```

```python
import os
from pathlib import Path

# Try to find cleaned_data.csv with multiple strategies
possible_paths = [
    Path("cleaned_data.csv"),  # Current directory
    Path("../cleaned_data.csv"),  # Parent directory
    Path("../../cleaned_data.csv"),  # Two levels up
    Path("../data/cleaned_data.csv"),  # data subdirectory
    Path("data/cleaned_data.csv"),  # data in current directory
]

csv_path = None
for path in possible_paths:
    if path.exists():
        csv_path = path
        print(f"✅ Found data file at: {path.absolute()}")
        break

if csv_path is None:
    print("❌ Error: cleaned_data.csv not found!")
    print(f"Current working directory: {os.getcwd()}")
    print(f"Available files: {os.listdir('.')}")
    raise FileNotFoundError("cleaned_data.csv not found in any expected location")

try:
    df = pd.read_csv(csv_path)
    print(f"\n✅ Data loaded successfully!")
    print(f"Shape: {df.shape}")
    print(f"Columns: {list(df.columns)}\n")
    df.head()
except Exception as e:
    print(f"❌ Error reading CSV: {e}")
    raise
```

```python
df.info()
df.describe()
```

```python
plt.figure()
plt.hist(df["Price_in_Lakhs"], bins=50)
plt.title("Distribution of Property Prices")
plt.xlabel("Price (Lakhs)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure()
plt.hist(df["Size_in_SqFt"], bins=50)
plt.title("Distribution of Property Sizes")
plt.xlabel("Size (SqFt)")
plt.ylabel("Frequency")
plt.show()
```

```python
df.boxplot(column="Price_per_SqFt", by="Property_Type")
plt.title("Price per SqFt by Property Type")
plt.suptitle("")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Size_in_SqFt"], df["Price_in_Lakhs"])
plt.xlabel("Size (SqFt)")
plt.ylabel("Price (Lakhs)")
plt.title("Size vs Price")
plt.show()
```

```python
df.boxplot(column=["Price_per_SqFt", "Size_in_SqFt"])
plt.title("Outlier Detection")
plt.show()
```

```python
state_avg = df.groupby("State")["Price_per_SqFt"].mean().sort_values()

state_avg.plot(kind="bar")
plt.title("Avg Price per SqFt by State")
plt.show()
```

```python
city_avg = df.groupby("City")["Price_in_Lakhs"].mean().sort_values(ascending=False)

city_avg.head(10).plot(kind="bar")
plt.title("Top Cities by Average Price")
plt.show()
```

```python
locality_age = df.groupby("Locality")["Age_of_Property"].median()

locality_age.head(10).plot(kind="bar")
plt.title("Median Property Age by Locality")
plt.show()
```

```python
pd.crosstab(df["City"], df["BHK"]).plot(kind="bar", stacked=True)
plt.title("BHK Distribution Across Cities")
plt.show()
```

```python
top_localities = df.groupby("Locality")["Price_in_Lakhs"].mean().sort_values(ascending=False).head(5)

top_localities.plot(kind="bar")
plt.title("Top 5 Expensive Localities")
plt.show()
```

```python
corr = df.select_dtypes(include=np.number).corr()

plt.figure()
plt.imshow(corr)
plt.colorbar()
plt.title("Correlation Matrix")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Nearby_Schools"], df["Price_per_SqFt"])
plt.xlabel("Nearby Schools")
plt.ylabel("Price per SqFt")
plt.title("Schools vs Price")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Nearby_Hospitals"], df["Price_per_SqFt"])
plt.xlabel("Nearby Hospitals")
plt.ylabel("Price per SqFt")
plt.title("Hospitals vs Price")
plt.show()
```

```python
df.boxplot(column="Price_in_Lakhs", by="Furnished_Status")
plt.title("Price by Furnished Status")
plt.suptitle("")
plt.show()
```

```python
df.boxplot(column="Price_per_SqFt", by="Facing")
plt.title("Price per SqFt by Facing Direction")
plt.suptitle("")
plt.show()
```

```python
df["Owner_Type"].value_counts().plot(kind="bar")
plt.title("Owner Type Distribution")
plt.show()
```

```python
df["Availability_Status"].value_counts().plot(kind="bar")
plt.title("Availability Status Distribution")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Parking_Space"], df["Price_in_Lakhs"])
plt.xlabel("Parking Spaces")
plt.ylabel("Price")
plt.title("Parking vs Price")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Amenity_Score"], df["Price_per_SqFt"])
plt.xlabel("Amenity Score")
plt.ylabel("Price per SqFt")
plt.title("Amenities vs Price")
plt.show()
```

```python
plt.figure()
plt.scatter(df["Public_Transport_Accessibility"], df["Price_per_SqFt"])
plt.xlabel("Transport Accessibility")
plt.ylabel("Price per SqFt")
plt.title("Transport vs Price")
plt.show()
```

## 🔍 Key Insights

- Property prices are right-skewed with high-value outliers.
- Larger properties generally have higher prices but not linearly.
- Infrastructure (schools, hospitals, transport) positively impacts price.
- Amenities and furnishing increase property value.
- Location is a dominant factor in determining price and investment potential.
